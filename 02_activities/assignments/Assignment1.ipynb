```python
#importing libraries
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import PIL
import requests
import scipy
import matplotlib
```

```python
#Creating a graph on Python
np.random.seed(613)
x= np.arange(50)
y=np.random.randint(0,100,50)

fig, ax = plt.subplots(figsize = (6,4))
#ax.plot(x,y)
#ax.bar(x,y)
#ax.scatter(x,y)
ax.hist(y)
```

```python
#Adding title and label to X and Y axis
font1 = {'family': 'sans-serif','color':'blue','size':20}
font2 = {'family': 'monospace','color':'red','size':14}
fig, ax = plt.subplots(figsize=(5,4))
ax.plot(x,y)

ax.set_title('Total growth over time',fontdict=font1, loc='left')
ax.set_ylabel ('Total Growth',fontdict=font2, loc='center')
ax.set_xlabel ('Years since start',fontdict=font2, loc='center')
plt.show()
```

```python
#Adding grid and customizing it to the graph created above
fig, ax = plt.subplots(figsize=(5,4))
ax.plot(x,y,
           marker ='*',
           color ='green',
           linestyle ='--',
           linewidth =2,
           markersize=12,
           markeredgecolor ='#fa9369',
           markerfacecolor = '#000000')
ax.grid(color='gray',linestyle='--',linewidth=2)
```

```python
#Reproducing a heatmap from Python graph gallery
import seaborn as sns
import pandas as pd
from matplotlib import pyplot as plt
 
# Data set
url = 'https://raw.githubusercontent.com/holtzy/The-Python-Graph-Gallery/master/static/data/mtcars.csv'
df = pd.read_csv(url)
df = df.set_index('model')
 
# Prepare a vector of color mapped to the 'cyl' column
my_palette = dict(zip(df.cyl.unique(), ["orange","yellow","brown"]))
row_colors = df.cyl.map(my_palette)
 
# plot
sns.clustermap(df, metric="correlation", method="single", cmap="Blues", standard_scale=1, row_colors=row_colors)
plt.show()
```

```python
#Creating multiple line plots in one with legend
np.random.seed(613)
x= np.arange(50)
y1= np.random.randint(0,100,50)
y2= np.random.randint(0,100,50)
fig, ax = plt.subplots(figsize=(5,4))
ax.plot(x,y1,label='Person1')
ax.plot(x,y2,label='Person2')
ax.legend(loc = 'lower right', 
          frameon = True,
          fontsize=14,
          ncol=2,
          shadow = True)
```

```python
#trying different layouts
plt.style.use('default')
x1=np.arange(50)
y1=np.random.randint(0,75,50)
x2=np.array (["Luffy","Zoro","Sanji"])
y2= np.array([110,180,240])
```

```python
fig,(ax1,ax2)=plt.subplots(ncols=2,nrows=1,figsize=(7,3))
ax1.scatter(x1,y1)
ax2.bar(x2,y2)
plt.xlabel("Total Sales",fontdict=font3, color="blue", loc = "left")
ax1.grid(color='gray', linestyle='--')
ax2.grid(color='gray', linestyle='--')
ax1.annotate("This is important!", xy=(10,70), xytext= (20,94),
            arrowprops = dict(facecolor = "black", shrink = 0.05))
```

```python
fig, someaxes = plt.subplot_mosaic([['ax1','ax3'],
                                    ['ax2','ax3']],
                                     figsize=(5,4),
                                     layout="constrained")
someaxes["ax1"].scatter(x1,y1)
someaxes["ax2"].bar(x2,y2)
someaxes["ax3"].plot(x1,y1)
someaxes["ax1"].set_xlabel('A big Label', fontsize=18)
someaxes["ax2"].set_xlabel('A label', fontsize=18)
someaxes["ax3"].set_xlabel('another', fontsize=18)

```
