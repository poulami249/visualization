```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
#Reading the CSV file
outbreak = pd.read_csv("C:/Users/poula/visualization/02_activities/assignments/ob_report_2024.csv")

#Renaming the Outbreak setting types
outbreak["Outbreak Setting"]=outbreak["Outbreak Setting"].replace({"Retirement Home":"RH","Hospital-Chronic Care":"HCC","Hospital-Acute Care":"HAC","Hospital-Psychiatric":"HP","Transitional Care":"TC"})
outbreak.info()
```

```python
#Counting the number of outbreaks by each setting and type of outbreak
outbreak_setting = outbreak["Outbreak Setting"].value_counts()
outbreak_type=outbreak["Type of Outbreak"].value_counts()
```

```python
#Creating 2 side-by-side bar charts using Matplotlib
fig,(ax1,ax2)=plt.subplots(ncols=2,nrows=1,figsize=(26,7))
ax1.bar(outbreak_setting.index,outbreak_setting.values,color="blue")
font3={'family':'serif','color':'black'}
ax1.set_ylabel("Number of Incidents",fontsize=22)
ax1.set_xlabel("Outbreak Setting",fontsize=22)
ax1.set_title("Number of Outbreak Incidents by Setting",fontdict=font3, color="black",loc='left',fontweight="bold",fontsize=24)
ax2.bar(outbreak_type.index,outbreak_type.values,color="blue")
ax2.set_ylabel("Number of Incidents",fontsize=22)
ax2.set_xlabel("Outbreak Type",fontsize=22)
ax2.set_title("Number of Outbreak Incidents by Type",fontdict=font3, color="black",loc='left',fontweight="bold",fontsize=24)

#Adding grid to the bar charts
ax1.grid(color='gray', linestyle='--')
ax2.grid(color='gray', linestyle='--')

#Customizing the X-axis values
ax1.tick_params(axis="x", labelsize=19)
ax2.tick_params(axis="x", labelsize=19)

#Creating a box to display the full form of each Outbreak setting type as they were renamed before
text_annotations = [
    "LTCH : Long-Term Care Home",
    "RH : Retirement Home",
    "HCC : Hospital-Chronic Care",
    'HAC : Hospital-Acute Care',
    'HP : Hospital-Psychiatric',
    'TC : Transitional Care'
]
combined_text = "\n".join(text_annotations)
text = ax1.text(3.25,614, combined_text, fontsize=15, color='black', ha='left', va='top')
text.set_bbox(dict(facecolor='white', edgecolor='black', boxstyle='round,pad=1'))

#Adding a Alt-text to describe the visualization
alt_text = "Two bar charts illustrating the number of outbreak incidents in the City of Toronto in 2024, categorized by outbreak setting type and infection type respectively. The data reveals that Long-Term Care Homes are the most affected setting, with Respiratory infections being the most prevalent."
plt.figtext(0.5, -0.1, alt_text, wrap=True, horizontalalignment='center', fontsize=22)
plt.show()
```
